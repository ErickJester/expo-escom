# Celda 3

```python
# ─── Esta celda NO entrena nada, solo revisa qué hay en Drive ───

def count_imgs_in_dir(dir_path):
    """Cuenta imágenes en una carpeta incluyendo subdirectorios."""
    count = 0
    for _, _, files in os.walk(dir_path):
        count += sum(
            1 for f in files
            if f.lower().endswith(('.jpg', '.jpeg', '.png', '.webp'))
        )
    return count

print("═" * 60)
print("📁 DIAGNÓSTICO DEL DATASET")
print("═" * 60)

CLASSES = []   # Se llenará solo con las que tengan imágenes

for cls in ALL_CLASSES:
    cls_dir = os.path.join(DATASET_ROOT, cls)

    if not os.path.exists(cls_dir):
        print(f"  ⬜ {cls:22s} → no existe (se saltará)")
        continue

    n     = count_imgs_in_dir(cls_dir)
    limite = MAX_OTRA if cls == 'otra' else MAX_PER_CLASS

    if n == 0:
        print(f"  ⚠️  {cls:22s} → carpeta vacía (se saltará)")
    else:
        uso = min(n, limite)
        CLASSES.append(cls)
        sufijo = " (con subcarpetas)" if cls == 'otra' else ""
        print(f"  ✅ {cls:22s} → {n:5,} imgs disponibles, usará {uso}{sufijo}")

NUM_CLASSES = len(CLASSES)
print(f"\n{'═'*60}")
print(f"Clases activas : {NUM_CLASSES}")
print(f"Lista          : {CLASSES}")
print(f"Total estimado : ~{MAX_PER_CLASS*(NUM_CLASSES-1) + MAX_OTRA:,} imágenes")
```

# Celda 4

```python
class CartoonDataset(Dataset):
    def __init__(self, samples, transform=None):
        self.samples   = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, label = self.samples[idx]
        try:
            img = Image.open(path).convert('RGB')
        except Exception:
            # Si la imagen está corrupta → imagen gris de reemplazo
            img = Image.new('RGB', (224, 224), (128, 128, 128))
        if self.transform:
            img = self.transform(img)
        return img, torch.tensor(label, dtype=torch.float32)


def build_samples(root_dir, classes, max_per_class, max_otra):
    """
    Construye lista de (ruta_imagen, vector_one_hot).
    Funciona tanto para carpetas normales (pokemon/) como
    para 'otra' que tiene subcarpetas (otra/cartoons_anime/, etc.)
    os.walk maneja ambos casos de forma transparente.
    """
    all_samples = []

    for cls_idx, cls_name in enumerate(classes):
        cls_dir = os.path.join(root_dir, cls_name)
        if not os.path.exists(cls_dir):
            continue

        limite = max_otra if cls_name == 'otra' else max_per_class

        # os.walk recorre tanto archivos directos como subcarpetas
        imgs = []
        for root, _, files in os.walk(cls_dir):
            for f in files:
                if f.lower().endswith(('.jpg', '.jpeg', '.png', '.webp')):
                    imgs.append(os.path.join(root, f))

        # Mezclar ANTES de cortar → diversidad de subcarpetas en 'otra'
        np.random.shuffle(imgs)
        imgs = imgs[:limite]

        if not imgs:
            print(f"  ⚠️  {cls_name}: no se encontraron imágenes")
            continue

        # Vector one-hot: solo esta clase es 1, el resto 0
        label = np.zeros(len(classes), dtype=np.float32)
        label[cls_idx] = 1.0

        for path in imgs:
            all_samples.append((path, label.copy()))

        print(f"  ✅ {cls_name:22s}: {len(imgs):,} imágenes")

    # Mezclar el dataset completo
    np.random.seed(42)
    np.random.shuffle(all_samples)
    return all_samples


print("Construyendo dataset...")
all_samples = build_samples(DATASET_ROOT, CLASSES, MAX_PER_CLASS, MAX_OTRA)

# Split 80% train / 20% val
split_idx     = int(0.8 * len(all_samples))
train_samples = all_samples[:split_idx]
val_samples   = all_samples[split_idx:]

print(f"\n{'═'*50}")
print(f"Total  : {len(all_samples):,}")
print(f"Train  : {len(train_samples):,}  ({split_idx/len(all_samples):.0%})")
print(f"Val    : {len(val_samples):,}  ({(len(all_samples)-split_idx)/len(all_samples):.0%})")
```

# Celda 5


```python
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

# Train: con augmentación
transform_train = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(0.5),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.1),
    transforms.RandomRotation(15),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

# Val: sin augmentación (solo resize + normalize)
transform_val = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

train_ds = CartoonDataset(train_samples, transform_train)
val_ds   = CartoonDataset(val_samples,   transform_val)

# num_workers=2 es el límite seguro en Colab gratis
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                          num_workers=0, pin_memory=False)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=0, pin_memory=False)

print(f"✅ DataLoaders listos")
print(f"   Train  : {len(train_ds):,} imgs → {len(train_loader)} batches/época")
print(f"   Val    : {len(val_ds):,} imgs → {len(val_loader)} batches/época")
est_tiempo = len(train_loader) * 0.07  # ~70ms por batch en T4
print(f"   Tiempo estimado por época: ~{est_tiempo:.0f}s")
```

# Celda 6

```python
class CartoonClassifier(nn.Module):
    """
    MobileNetV2 + cabeza multi-etiqueta.
    num_classes se define dinámicamente según lo que hay en el dataset.
    Ahora: 8 clases. Cuando estén las 15, se reentrena con num_classes=15.
    """
    def __init__(self, num_classes):
        super().__init__()
        backbone      = models.mobilenet_v2(weights=models.MobileNet_V2_Weights.DEFAULT)
        self.features = backbone.features          # Salida: [B, 1280, 7, 7]
        self.avgpool  = nn.AdaptiveAvgPool2d((1, 1))

        # Cabeza: Linear → ReLU → Dropout → Linear(num_classes)
        self.head = nn.Sequential(
            nn.Linear(1280, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(512, num_classes),
            # SIN Sigmoid aquí → BCEWithLogitsLoss la aplica internamente
        )
        # Fase 1: solo cabeza entrenable
        for p in self.features.parameters():
            p.requires_grad = False

    def forward(self, x):
        x = self.features(x)       # [B, 1280, 7, 7]
        x = self.avgpool(x)        # [B, 1280, 1, 1]
        return self.head(torch.flatten(x, 1))  # [B, num_classes] → logits

    def unfreeze_last_n(self, n=3):
        for layer in list(self.features.children())[-n:]:
            for p in layer.parameters():
                p.requires_grad = True
        t = sum(p.numel() for p in self.parameters() if p.requires_grad)
        print(f"   🔥 {n} capas descongeladas → {t:,} params entrenables")


model = CartoonClassifier(num_classes=NUM_CLASSES).to(DEVICE)

total = sum(p.numel() for p in model.parameters())
train = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"✅ Modelo listo — {NUM_CLASSES} clases: {CLASSES}")
print(f"   Params totales     : {total:,}")
print(f"   Params Fase 1 (P1) : {train:,}  ({100*train/total:.1f}%)")
```

# Celda 7

```python
criterion = nn.BCEWithLogitsLoss()  # Sigmoid + BCE en una sola op, numéricamente estable


def compute_metrics(y_true, y_logits, threshold=THRESHOLD):
    """Macro F1 y mAP dado arrays numpy [N, C]."""
    probs  = 1 / (1 + np.exp(-y_logits))        # sigmoid manual (no requiere torch)
    y_pred = (probs >= threshold).astype(int)
    f1     = f1_score(y_true, y_pred, average='macro', zero_division=0)
    aps    = [
        average_precision_score(y_true[:, c], probs[:, c])
        for c in range(y_true.shape[1]) if y_true[:, c].sum() > 0
    ]
    return f1, (np.mean(aps) if aps else 0.0)


def train_epoch(model, loader, optimizer):
    model.train()
    total_loss = 0.0
    for i, (imgs, labels) in enumerate(loader):      # ← enumerate aquí
        imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        loss = criterion(model(imgs), labels)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
        if (i + 1) % 10 == 0:
            print(f"  batch {i+1}/{len(loader)}", end='\r')
    return total_loss / len(loader)


@torch.no_grad()
def eval_epoch(model, loader):
    model.eval()
    total_loss, yt_all, yl_all = 0.0, [], []
    for imgs, labels in loader:
        imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
        logits = model(imgs)
        total_loss += criterion(logits, labels).item()
        yt_all.append(labels.cpu().numpy())
        yl_all.append(logits.cpu().numpy())
    y_true, y_logits = np.concatenate(yt_all), np.concatenate(yl_all)
    f1, mAP = compute_metrics(y_true, y_logits)
    return total_loss / len(loader), f1, mAP


def save_best(state_dict, epoch, best_f1, phase, history):
    """
    Guarda el MEJOR modelo en Drive después de cada época.
    Si Colab se desconecta, el mejor checkpoint siempre está a salvo.
    """
    torch.save({
        'model_state_dict': state_dict,
        'classes'         : CLASSES,
        'num_classes'     : NUM_CLASSES,
        'best_f1'         : best_f1,
        'epoch'           : epoch,
        'phase'           : phase,
        'history'         : history,
        'threshold'       : THRESHOLD,
    }, SAVE_PATH)


print("✅ Funciones listas")
```

# Celda 8

```python
print("═" * 60)
print("🚀 FASE 1 — Backbone congelado | Solo entrena la cabeza")
print(f"   LR={LR_PHASE1} | Batch={BATCH_SIZE} | Épocas={NUM_EPOCHS_P1}")
print("═" * 60)

optimizer_p1 = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=LR_PHASE1, weight_decay=1e-4
)
scheduler_p1 = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer_p1, mode='max', factor=0.5, patience=2
)

history    = {'train_loss': [], 'val_loss': [], 'val_f1': [], 'val_mAP': []}
best_f1    = 0.0
best_state = None

for epoch in range(1, NUM_EPOCHS_P1 + 1):
    t0         = time.time()
    train_loss = train_epoch(model, train_loader, optimizer_p1)
    val_loss, val_f1, val_mAP = eval_epoch(model, val_loader)
    scheduler_p1.step(val_f1)
    elapsed    = time.time() - t0

    for k, v in zip(['train_loss','val_loss','val_f1','val_mAP'],
                    [train_loss, val_loss, val_f1, val_mAP]):
        history[k].append(v)

    star = " ⭐ guardado" if val_f1 > best_f1 else ""
    print(f"[P1 {epoch:02d}/{NUM_EPOCHS_P1}] {elapsed:.0f}s | "
          f"Loss {train_loss:.4f}/{val_loss:.4f} | "
          f"F1 {val_f1:.4f} | mAP {val_mAP:.4f}{star}")

    if val_f1 > best_f1:
        best_f1    = val_f1
        best_state = {k: v.clone() for k, v in model.state_dict().items()}
        save_best(best_state, epoch, best_f1, 'P1', history)

print(f"\n🏁 Mejor F1 Fase 1: {best_f1:.4f}")
```
