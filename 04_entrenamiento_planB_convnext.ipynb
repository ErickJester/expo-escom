# ExpoEscom — Plan B v0.1.0 · ConvNeXt-Small (robusto)

**Rol en la cascada:** segunda etapa. El Plan A (MobileNetV2) resuelve los casos fáciles (~80% del tráfico); lo que A **no** resuelve con confianza se reenvía aquí.

**Qué cambia respecto al Plan A**
- **Backbone:** ConvNeXt-Small (~50M params) en vez de MobileNetV2 (~3.4M).
- **Loss:** Focal Loss (alpha=0.25, gamma=2.0) — maneja el desbalance de `'otra'` automáticamente, sin `pos_weight` manual.
- **MixUp (50% de los batches):** fronteras suaves marca↔`'otra'`. Es la especialidad del Plan B.
- **Fine-tuning:** últimos 2 stages, LR 5e-5.

**Mismo dataset, mismo split que el Plan A** (seed 42, misma expansión de `'otra'`, misma validación capada) para que los dos modelos sean comparables y la cascada coherente.

**Infra Pro+:** reutiliza las 3 palancas del v3 — feature caching en Fase 1, fine-tuning sobre subconjunto balanceado, validación capada — más AMP / channels_last / TF32.

## Celda 4 — Diagnóstico del dataset (local)

Cuenta imágenes por clase ya en local. **`'otra'` se expande** en sus subcarpetas (cartoons_anime / noise / real_life), igual que el modelo local y que el Plan A.

```python
def detect_classes(root):
    """Subcarpetas con imágenes = clases. 'otra' se expande en sus
    subcarpetas; las clases de ruido van al final."""
    cartoons, noise = [], []
    for name in sorted(os.listdir(root)):
        d = os.path.join(root, name)
        if not os.path.isdir(d):
            continue
        if is_noise(name):
            subs = [s for s in sorted(os.listdir(d))
                    if os.path.isdir(os.path.join(d, s))]
            noise += [f"{name}/{s}" for s in subs] if subs else [name]
        else:
            cartoons.append(name)
    return cartoons + noise


def count_imgs_in_dir(dir_path):
    n = 0
    for _, _, files in os.walk(dir_path):
        n += sum(1 for f in files if f.lower().endswith(('.jpg', '.jpeg', '.png', '.webp')))
    return n


print("═" * 60)
print("📁 DIAGNÓSTICO DEL DATASET (local)")
print("═" * 60)

CLASSES = []
for cls in detect_classes(DATASET_ROOT):
    cls_dir = os.path.join(DATASET_ROOT, *cls.split('/'))
    n       = count_imgs_in_dir(cls_dir)
    limite  = MAX_OTRA if is_noise(cls) else MAX_PER_CLASS
    if n == 0:
        print(f"  ⚠️  {cls:<22s} → carpeta vacía (se salta)")
        continue
    CLASSES.append(cls)
    print(f"  ✅ {cls:<22s} → {n:7,} imgs, usará {min(n, limite):,}")

NUM_CLASSES = len(CLASSES)
print(f"\n{'═'*60}")
print(f"Clases activas : {NUM_CLASSES}")
print(f"Lista          : {CLASSES}")
```

## Celda 5 — Dataset + recolección de muestras

```python
class CartoonDataset(Dataset):
    def __init__(self, samples, transform=None):
        self.samples   = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, label = self.samples[idx]
        try:
            img = Image.open(path).convert('RGB')
        except Exception:
            img = Image.new('RGB', (224, 224), (128, 128, 128))
        if self.transform:
            img = self.transform(img)
        return img, torch.tensor(label, dtype=torch.long)


def build_samples(root_dir, classes, max_per_class, max_otra):
    all_samples = []
    bar = tqdm(list(enumerate(classes)), desc='Recolectando', unit='clase')
    for cls_idx, cls_name in bar:
        cls_dir = os.path.join(root_dir, *cls_name.split('/'))
        if not os.path.exists(cls_dir):
            continue
        limite = max_otra if is_noise(cls_name) else max_per_class
        imgs = []
        for root, _, files in os.walk(cls_dir):
            for f in files:
                if f.lower().endswith(('.jpg', '.jpeg', '.png', '.webp')):
                    imgs.append(os.path.join(root, f))
        np.random.shuffle(imgs)
        imgs = imgs[:limite]
        if not imgs:
            continue
        for path in imgs:
            all_samples.append((path, cls_idx))
        bar.set_postfix_str(f"{cls_name}: {len(imgs):,}")
    return all_samples


print("Construyendo lista de muestras...")
np.random.seed(42)
all_samples = build_samples(DATASET_ROOT, CLASSES, MAX_PER_CLASS, MAX_OTRA)
print(f"\nTotal: {len(all_samples):,} imágenes")
```

## Celda 6 — Split (validación capada por clase, seed 42)

```python
by_class = defaultdict(list)
for s in all_samples:
    by_class[s[1]].append(s)

rng = np.random.default_rng(42)
train_samples, val_samples = [], []
for c, items in by_class.items():
    items = list(items)
    rng.shuffle(items)
    val_samples   += items[:VAL_PER_CLASS]
    train_samples += items[VAL_PER_CLASS:]

rng.shuffle(train_samples)
train_labels = np.array([s[1] for s in train_samples])
class_counts = np.bincount(train_labels, minlength=NUM_CLASSES).astype(float)

print(f"Train: {len(train_samples):,}  |  Val: {len(val_samples):,}")
print("\nImágenes de train por clase:")
for i, cls in enumerate(CLASSES):
    print(f"  {cls:22s}: {int(class_counts[i]):,}")
```

## Celda 7 — Transforms y DataLoaders (Plan B)

Aumentación "lo mismo del Plan A" (crop, flip, jitter, erasing). **MixUp NO va aquí**: se aplica en el bucle de entrenamiento, al 50% de los batches.

```python
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

# Aumentación Fase 2 (imágenes reales). MixUp se aplica aparte en el loop.
transform_train = transforms.Compose([
    transforms.RandomResizedCrop(224, scale=(0.6, 1.0)),
    transforms.RandomHorizontalFlip(0.5),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.1),
    transforms.RandomRotation(15),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    transforms.RandomErasing(p=0.25),
])

# Sin aumentación: para extraer features y para validar
transform_plain = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

loader_kwargs = dict(num_workers=NUM_WORKERS, pin_memory=True,
                     persistent_workers=(NUM_WORKERS > 0))
if NUM_WORKERS > 0:
    loader_kwargs['prefetch_factor'] = 4

# Loader para EXTRAER features de todo el train (sin aumentación, sin shuffle)
feat_loader = DataLoader(CartoonDataset(train_samples, transform_plain),
                         batch_size=BATCH_FEAT, shuffle=False, **loader_kwargs)

# Loader de validación con imágenes reales (Fase 2 y reporte final)
val_loader  = DataLoader(CartoonDataset(val_samples, transform_plain),
                         batch_size=BATCH_FEAT, shuffle=False, **loader_kwargs)

print(f"✅ Loaders | feat: {len(feat_loader)} batches | val: {len(val_loader)} batches")
```

## Celda 8 — Modelo: ConvNeXt-Small + cabeza propia

La `LayerNorm` final del backbone se mantiene como parte del extractor para que las features cacheadas en Fase 1 sean deterministas (igual que el truco del backbone congelado del v3).

```python
class CartoonConvNeXt(nn.Module):
    """ConvNeXt-Small pre-entrenado + cabeza propia."""

    def __init__(self, num_classes):
        super().__init__()
        backbone      = models.convnext_small(weights=models.ConvNeXt_Small_Weights.DEFAULT)
        self.features = backbone.features          # stem + 4 stages → [B, 768, 7, 7]
        self.avgpool  = backbone.avgpool           # AdaptiveAvgPool2d(1)
        self.norm     = backbone.classifier[0]     # LayerNorm2d(768)
        self.head = nn.Sequential(
            nn.Linear(768, 512),
            nn.GELU(),
            nn.Dropout(0.4),
            nn.Linear(512, num_classes),
        )
        # Fase 1: todo el backbone congelado (incl. la norm final)
        for p in self.features.parameters():
            p.requires_grad = False
        for p in self.norm.parameters():
            p.requires_grad = False

    def extract(self, x):
        x = self.features(x)
        x = self.avgpool(x)
        x = self.norm(x)
        return torch.flatten(x, 1)      # [B, 768]

    def forward(self, x):
        return self.head(self.extract(x))

    def unfreeze_last_n(self, n=3):
        """Descongela las últimas n 'children' de features (los stages
        superiores) + la LayerNorm final. n=3 ≈ últimos 2 stages."""
        for layer in list(self.features.children())[-n:]:
            for p in layer.parameters():
                p.requires_grad = True
        for p in self.norm.parameters():
            p.requires_grad = True
        t = sum(p.numel() for p in self.parameters() if p.requires_grad)
        print(f"   🔥 últimas {n} children + norm descongeladas → {t:,} params entrenables")


def save_best(state_dict, epoch, best_f1, phase, history):
    torch.save({
        'model_state_dict': state_dict,
        'classes'         : CLASSES,
        'num_classes'     : NUM_CLASSES,
        'best_f1'         : best_f1,
        'epoch'           : epoch,
        'phase'           : phase,
        'history'         : history,
        'arch'            : 'convnext_small',
        'plan'            : 'B',
        'version'         : VERSION,
    }, SAVE_PATH)


model = CartoonConvNeXt(NUM_CLASSES).to(DEVICE).to(memory_format=torch.channels_last)
total     = sum(p.numel() for p in model.parameters())
trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"✅ ConvNeXt-Small listo — {NUM_CLASSES} clases | {total:,} params "
      f"({trainable:,} entrenables en Fase 1)")
```

## Celda 9 — Focal Loss, MixUp y funciones (features / train / eval)

```python
import torch.nn.functional as F

class FocalLoss(nn.Module):
    """Focal Loss multiclase (softmax). Reduce el peso de los ejemplos
    fáciles (el exceso de 'otra') → maneja el desbalance sin pos_weight."""
    def __init__(self, alpha=0.25, gamma=2.0, weight=None):
        super().__init__()
        self.alpha, self.gamma, self.weight = alpha, gamma, weight

    def forward(self, logits, target):
        ce = F.cross_entropy(logits, target, weight=self.weight, reduction='none')
        pt = torch.exp(-ce)
        return (self.alpha * (1 - pt) ** self.gamma * ce).mean()


def mixup_batch(x, y, beta, prob):
    """Con prob, mezcla el batch con una permutación de sí mismo.
    Devuelve (x_mezclado, y_a, y_b, lam). lam=1 → sin mezcla."""
    if np.random.rand() < prob:
        lam = float(np.random.beta(beta, beta))
        idx = torch.randperm(x.size(0), device=x.device)
        x   = lam * x + (1 - lam) * x[idx]
        return x, y, y[idx], lam
    return x, y, y, 1.0


def mixup_loss(criterion, logits, y_a, y_b, lam):
    return lam * criterion(logits, y_a) + (1 - lam) * criterion(logits, y_b)


criterion = FocalLoss(FOCAL_ALPHA, FOCAL_GAMMA).to(DEVICE)
scaler    = torch.amp.GradScaler('cuda', enabled=USE_AMP)


def compute_metrics(y_true, y_pred):
    f1m = f1_score(y_true, y_pred, average='macro', zero_division=0)
    f1c = f1_score(y_true, y_pred, average=None,    zero_division=0)
    return f1m, f1c


@torch.no_grad()
def extract_features(model, loader, desc=''):
    # Una pasada por el backbone congelado -> vectores [N, 768] en fp16
    model.eval()
    feats, labs = [], []
    for imgs, labels in tqdm(loader, desc=f'features {desc}', unit='batch'):
        imgs = imgs.to(DEVICE, non_blocking=True).to(memory_format=torch.channels_last)
        with torch.autocast('cuda', dtype=torch.float16, enabled=USE_AMP):
            f = model.extract(imgs)
        feats.append(f.half().cpu())
        labs.append(labels)
    return torch.cat(feats), torch.cat(labs)


def train_epoch(model, loader, optimizer, desc='train'):
    model.train()
    total = 0.0
    bar = tqdm(loader, desc=desc, unit='batch', leave=False)
    for imgs, labels in bar:
        imgs   = imgs.to(DEVICE, non_blocking=True).to(memory_format=torch.channels_last)
        labels = labels.to(DEVICE, non_blocking=True)
        imgs, ya, yb, lam = mixup_batch(imgs, labels, MIXUP_BETA, MIXUP_PROB)
        optimizer.zero_grad()
        with torch.autocast('cuda', dtype=torch.float16, enabled=USE_AMP):
            logits = model(imgs)
            loss   = mixup_loss(criterion, logits, ya, yb, lam)
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        scaler.step(optimizer)
        scaler.update()
        total += loss.item()
        bar.set_postfix(loss=f"{loss.item():.4f}")
    return total / len(loader)


@torch.no_grad()
def eval_epoch(model, loader, desc='val'):
    model.eval()
    total, gts, preds = 0.0, [], []
    for imgs, labels in tqdm(loader, desc=desc, unit='batch', leave=False):
        imgs   = imgs.to(DEVICE, non_blocking=True).to(memory_format=torch.channels_last)
        labels = labels.to(DEVICE, non_blocking=True)
        with torch.autocast('cuda', dtype=torch.float16, enabled=USE_AMP):
            logits = model(imgs)
            total += criterion(logits, labels).item()
        preds.append(logits.argmax(1).cpu().numpy())
        gts.append(labels.cpu().numpy())
    y_true, y_pred = np.concatenate(gts), np.concatenate(preds)
    f1m, f1c = compute_metrics(y_true, y_pred)
    return total / len(loader), f1m, f1c


print("✅ Focal Loss + MixUp + funciones listas")
```

## Celda 11 — Fase 1: cabeza sobre features cacheadas (Focal + MixUp)

MixUp también aquí: mezcla vectores de features (es lineal) → la cabeza aprende fronteras suaves desde el inicio.

```python
print("═" * 60)
print("🚀 FASE 1 — Cabeza sobre features cacheadas (Focal + MixUp)")
print("═" * 60)

feat_train_loader = DataLoader(TensorDataset(train_feats, train_feat_labels),
                               batch_size=HEAD_BATCH, shuffle=True)
feat_val_loader   = DataLoader(TensorDataset(val_feats, val_feat_labels),
                               batch_size=HEAD_BATCH, shuffle=False)

optimizer_p1 = torch.optim.AdamW(model.head.parameters(), lr=LR_PHASE1, weight_decay=1e-4)
scheduler_p1 = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer_p1, mode='max', factor=0.5, patience=2)

history    = {'train_loss': [], 'val_loss': [], 'val_f1': []}
best_f1, best_state = 0.0, None

for epoch in range(1, NUM_EPOCHS_P1 + 1):
    t0 = time.time()
    model.head.train()
    tl = 0.0
    bar = tqdm(feat_train_loader, desc=f'P1 {epoch:02d}/{NUM_EPOCHS_P1}', unit='batch', leave=False)
    for fb, lb in bar:
        fb, lb = fb.float().to(DEVICE), lb.to(DEVICE)
        fb, ya, yb, lam = mixup_batch(fb, lb, MIXUP_BETA, MIXUP_PROB)
        optimizer_p1.zero_grad()
        out  = model.head(fb)
        loss = mixup_loss(criterion, out, ya, yb, lam)
        loss.backward()
        optimizer_p1.step()
        tl += loss.item()
        bar.set_postfix(loss=f"{loss.item():.4f}")
    tl /= len(feat_train_loader)

    model.head.eval()
    vl, gts, preds = 0.0, [], []
    with torch.no_grad():
        for fb, lb in feat_val_loader:
            out = model.head(fb.float().to(DEVICE))
            vl += criterion(out, lb.to(DEVICE)).item()
            preds.append(out.argmax(1).cpu().numpy())
            gts.append(lb.numpy())
    vl /= len(feat_val_loader)
    val_f1 = f1_score(np.concatenate(gts), np.concatenate(preds),
                      average='macro', zero_division=0)
    scheduler_p1.step(val_f1)

    history['train_loss'].append(tl); history['val_loss'].append(vl)
    history['val_f1'].append(val_f1)
    star = " ⭐" if val_f1 > best_f1 else ""
    print(f"[P1 {epoch:02d}/{NUM_EPOCHS_P1}] {time.time()-t0:.0f}s | "
          f"loss {tl:.4f}/{vl:.4f} | F1 {val_f1:.4f}{star}")

    if val_f1 > best_f1:
        best_f1    = val_f1
        best_state = {k: v.clone() for k, v in model.state_dict().items()}
        save_best(best_state, epoch, best_f1, 'P1', history)

print(f"\n🏁 Mejor F1 Fase 1: {best_f1:.4f}")
```

## Celda 12 — Fase 2: fine-tuning últimos 2 stages (Focal + MixUp en imágenes reales)

```python
print("═" * 60)
print(f"🔥 FASE 2 — Fine-tuning últimos 2 stages | {FT_PER_CLASS:,} imgs/clase | MixUp")
print("═" * 60)

# Subconjunto balanceado de train (imágenes reales, con aumentación)
by_class_train = defaultdict(list)
for s in train_samples:
    by_class_train[s[1]].append(s)
rng2 = np.random.default_rng(0)
ft_samples = []
for c, items in by_class_train.items():
    items = list(items)
    rng2.shuffle(items)
    ft_samples += items[:FT_PER_CLASS]
rng2.shuffle(ft_samples)
print(f"Subconjunto fine-tuning: {len(ft_samples):,} imágenes")

ft_loader = DataLoader(CartoonDataset(ft_samples, transform_train),
                       batch_size=BATCH_FT, shuffle=True, **loader_kwargs)

model.load_state_dict(best_state)
model.unfreeze_last_n(UNFREEZE_N)
model = model.to(memory_format=torch.channels_last)

# torch.compile: ~20-40% speedup en A100/L4 (PyTorch 2.x)
# reduce-overhead = fusión de kernels sin el costo de max-autotune
if USE_AMP and hasattr(torch, 'compile'):
    print('   ⚡ torch.compile (reduce-overhead)...')
    model = torch.compile(model, mode='reduce-overhead')
    print('   ✅ modelo compilado')

optimizer_p2 = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=LR_PHASE2, weight_decay=1e-5)
scheduler_p2 = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer_p2, T_max=NUM_EPOCHS_P2)

for epoch in range(1, NUM_EPOCHS_P2 + 1):
    t0 = time.time()
    train_loss = train_epoch(model, ft_loader, optimizer_p2, desc=f'P2 {epoch:02d}/{NUM_EPOCHS_P2}')
    val_loss, val_f1, _ = eval_epoch(model, val_loader, desc='val')
    scheduler_p2.step()

    history['train_loss'].append(train_loss); history['val_loss'].append(val_loss)
    history['val_f1'].append(val_f1)
    star = " ⭐" if val_f1 > best_f1 else ""
    print(f"[P2 {epoch:02d}/{NUM_EPOCHS_P2}] {time.time()-t0:.0f}s | "
          f"loss {train_loss:.4f}/{val_loss:.4f} | F1 {val_f1:.4f}{star}")

    if val_f1 > best_f1:
        best_f1    = val_f1
        best_state = {k: v.clone() for k, v in model.state_dict().items()}
        save_best(best_state, epoch, best_f1, 'P2', history)

print(f"\n🏆 Mejor F1 Plan B: {best_f1:.4f}")
print(f"   Guardado en: {SAVE_PATH}")
```

## Celda 14 — Confianza del Plan B y cómo encaja en la cascada

```python
# La cascada vive en server.py: Plan A clasifica primero; si su softmax
# máximo < UMBRAL_A, la imagen se reenvía al Plan B. Aquí medimos qué tan
# seguro está el Plan B en aciertos vs errores, como referencia.
model.eval()
max_probs, correct = [], []
with torch.no_grad():
    for imgs, labels in tqdm(val_loader, desc='confianza', unit='batch'):
        imgs = imgs.to(DEVICE, non_blocking=True).to(memory_format=torch.channels_last)
        with torch.autocast('cuda', dtype=torch.float16, enabled=USE_AMP):
            probs = F.softmax(model(imgs), dim=1)
        p, pred = probs.max(1)
        max_probs.append(p.float().cpu().numpy())
        correct.append((pred.cpu() == labels).numpy())

max_probs = np.concatenate(max_probs)
correct   = np.concatenate(correct)
print(f"Confianza media en aciertos : {max_probs[correct].mean():.3f}")
print(f"Confianza media en errores  : {max_probs[~correct].mean():.3f}")
print(f"Cobertura @ conf>0.90       : {(max_probs > 0.90).mean()*100:.1f}% de val")

print("\n" + "═"*60)
print("🔗 CÓMO ENCAJA EN LA CASCADA (server.py)")
print("═"*60)
print("""
  1. Cargar AMBOS checkpoints al arrancar:
       plan_a = MobileNetV2  (cartoon_v3.pt)
       plan_b = ConvNeXt     (este .pt; usar el campo 'arch' para elegir la clase)
  2. Inferencia:
       probs_a = softmax(plan_a(img))
       if probs_a.max() >= UMBRAL_A:      # caso fácil  -> Plan A
           return clase_a
       else:                              # ambiguo     -> Plan B
           return argmax(plan_b(img))
  3. UMBRAL_A se calibra con la confianza del Plan A (no de aquí);
     un punto de partida razonable es 0.80-0.90.
""")
```
